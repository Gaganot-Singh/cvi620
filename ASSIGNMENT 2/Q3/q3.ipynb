# Q3: House Price Prediction


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import accuracy_score
```

# Load datasets

```python
# 1. Load CSV files
train_df = pd.read_csv("mnist_train.csv")
test_df = pd.read_csv("mnist_test.csv")
```

# Separate features and labels

```python
# 2. Split into features (X) and labels (y)
X_train = train_df.iloc[:, 1:].values  # pixel values (flattened)
y_train = train_df.iloc[:, 0].values   # digit label (0–9)

X_test = test_df.iloc[:, 1:].values
y_test = test_df.iloc[:, 0].values

print("Training samples:", X_train.shape[0])
print("Test samples:", X_test.shape[0])
```

# Normalize pixel values

```python
# 3. Normalize pixel values to 0–1 range
X_train = X_train / 255.0
X_test = X_test / 255.0
```

# Standard scaling

```python
# 4. Feature scaling using StandardScaler
sc = StandardScaler()
X_train_scaled = sc.fit_transform(X_train)
X_test_scaled = sc.transform(X_test)
```

# Train Logistic Regression

```python
# 5. Logistic Regression Model
print("\nTraining Logistic Regression...")
lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train, y_train)
lr_preds = lr_model.predict(X_test)
lr_acc = accuracy_score(y_test, lr_preds)
print("Logistic Regression Accuracy:", round(lr_acc * 100, 2), "%")
```

# Train KNN

```python
# 6. KNN Model
print("\nTraining K-Nearest Neighbors (KNN)...")
knn_model = KNeighborsClassifier(n_neighbors=3)
knn_model.fit(X_train, y_train)
knn_preds = knn_model.predict(X_test)
knn_acc = accuracy_score(y_test, knn_preds)
print("KNN Accuracy:", round(knn_acc * 100, 2), "%")
```

# Train SGD Classifier

```python
# 7. SGD Classifier
print("\nTraining SGD Classifier...")
sgd_model = SGDClassifier()
sgd_model.fit(X_train, y_train)
predictions = sgd_model.predict(X_test)
sgd_acc = accuracy_score(y_test, predictions)
print("SGD Accuracy:", round(sgd_acc * 100, 2), "%")
```

# Compare models

```python
# Compare all 3 models and select the best
accuracies = {
    "Logistic Regression": lr_acc,
    "SGD Classifier": sgd_acc,
    "KNN": knn_acc
}

best_model_name = max(accuracies, key=accuracies.get)
best_model = {
    "Logistic Regression": lr_model,
    "SGD Classifier": sgd_model,
    "KNN": knn_model
}[best_model_name]

results_df = pd.DataFrame({
    "Model": ["Logistic Regression", "K-Nearest Neighbors", "SGD Classifier"],
    "Accuracy": [lr_acc, sgd_acc, knn_acc]
})
print("\nModel Accuracy Comparison:")
print(results_df)
print(f"\nBest Model: {best_model_name} with {accuracies[best_model_name]*100:.2f}% accuracy")
```
